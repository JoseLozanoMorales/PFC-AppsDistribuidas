# Análisis reproducible del experimento 2PC frente a Saga

Este cuaderno lee los datos crudos versionados. La generación, el oráculo, los intervalos y las figuras oficiales se ejecutan con `run_paso8.py`; aquí se inspeccionan independientemente sus invariantes principales usando únicamente la biblioteca estándar de Python.

```python
from pathlib import Path
import csv

base = Path.cwd()
if not (base / 'resultados').exists():
    base = base / 'experiments' / 'paso8'
raw_path = base / 'resultados' / 'experimento_crudo.csv'
with raw_path.open(encoding='utf-8-sig', newline='') as stream:
    rows = list(csv.DictReader(stream))
len(rows), sorted({r['coord'] for r in rows}), sorted({int(r['concurrencia']) for r in rows})
```

```python
assert len(rows) == 120, 'Se esperaban 120 corridas'
assert all(r['oracle_pass'].lower() == 'true' for r in rows)
assert all(float(r['tasa_inconsistencia']) == 0.0 for r in rows)
print('Oráculo aprobado en 120/120 corridas; inconsistencias observadas: 0.')
```

```python
from statistics import median

resumen = {}
for coord in ('2pc', 'saga'):
    selected = [float(r['latencia_pago_p95_ms']) for r in rows if r['coord'] == coord]
    resumen[coord] = {'corridas': len(selected), 'mediana_p95_ms': round(median(selected), 3)}
resumen
```
